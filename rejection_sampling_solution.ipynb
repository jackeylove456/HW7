# 拒绝采样得到的样本分布证明设目标概率密度函数为 $f(x)$，易采样的提议分布密度为 $g(x)$，并取常数 $M\ge \sup_x rac{f(x)}{g(x)}$。拒绝采样过程为：1. 先从 $g(x)$ 采样 $Y$，再从均匀分布 $U\sim\mathrm{Unif}(0,1)$ 采样 $U$；2. 若 $U\le rac{f(Y)}{M g(Y)}$ 则接受并输出 $X=Y$，否则回到第 1 步。记接受事件为 $A=\{U\le rac{f(Y)}{M g(Y)}\}$。对可测集合 $B$ 有egin{aligned}\mathbb{P}(X\in B)&= \mathbb{P}(Y\in B\mid A) \&= rac{\mathbb{P}(Y\in B, A)}{\mathbb{P}(A)} \&= rac{\int_B \int_0^{f(y)/(M g(y))} g(y)\,\mathrm{d}u\,\mathrm{d}y}{\int_{\mathbb{R}} \int_0^{f(y)/(M g(y))} g(y)\,\mathrm{d}u\,\mathrm{d}y} \&= rac{\int_B rac{f(y)}{M}\,\mathrm{d}y}{\int_{\mathbb{R}} rac{f(y)}{M}\,\mathrm{d}y} \&= \int_B f(y)\,\mathrm{d}y,\end{aligned}因为 $f$ 是概率密度函数，$\int_{\mathbb{R}} f(y)\,\mathrm{d}y=1$。因此接受后的样本 $X$ 的概率密度正是目标密度 $f(x)$。接受概率为 $\mathbb{P}(A)=rac{1}{M}$，常用来评估拒绝采样的效率。

```python
import numpy as npfrom scipy.stats import beta, uniformrng = np.random.default_rng(0)# 目标分布 Beta(2,5)，提议分布为 U(0,1)M = 2.5  # 覆盖常数，使得 f(x) <= M g(x)num_samples = 100000accepted = []tries = 0while len(accepted) < num_samples:    y = rng.uniform(0, 1)    u = rng.uniform(0, 1)    if u <= beta.pdf(y, 2, 5) / (M * uniform.pdf(y, 0, 1)):        accepted.append(y)    tries += 1samples = np.array(accepted)print(f"接受率≈{len(samples)/tries:.3f} (理论值 1/M = {1/M:.3f})")print(f"样本均值≈{samples.mean():.4f} (理论均值={beta.mean(2,5):.4f})")
```
